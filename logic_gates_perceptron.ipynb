```python
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.neural_network import MLPClassifier

# --------------------------------
# Input Data
# --------------------------------

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])


# --------------------------------
# AND Gate
# --------------------------------

AND = np.array([0, 0, 0, 1])

and_model = Perceptron(
    max_iter=1000,
    random_state=42
)

and_model.fit(X, AND)

and_prediction = and_model.predict(X)

print("AND GATE")
print("Actual    :", AND)
print("Predicted :", and_prediction)


# --------------------------------
# OR Gate
# --------------------------------

OR = np.array([0, 1, 1, 1])

or_model = Perceptron(
    max_iter=1000,
    random_state=42
)

or_model.fit(X, OR)

or_prediction = or_model.predict(X)

print("\nOR GATE")
print("Actual    :", OR)
print("Predicted :", or_prediction)


# --------------------------------
# XOR - Single Layer Perceptron
# --------------------------------

XOR = np.array([0, 1, 1, 0])

xor_single = Perceptron(
    max_iter=1000,
    random_state=42
)

xor_single.fit(X, XOR)

xor_single_prediction = xor_single.predict(X)

print("\nXOR - SINGLE LAYER")
print("Actual    :", XOR)
print("Predicted :", xor_single_prediction)


# --------------------------------
# XOR - Multi Layer Perceptron
# --------------------------------

xor_multi = MLPClassifier(
    hidden_layer_sizes=(2,),
    activation='logistic',
    solver='lbfgs',
    max_iter=5000,
    random_state=42
)

xor_multi.fit(X, XOR)

xor_multi_prediction = xor_multi.predict(X)

print("\nXOR - MULTI LAYER")
print("Actual    :", XOR)
print("Predicted :", xor_multi_prediction)

# Accuracy
accuracy = np.mean(xor_multi_prediction == XOR) * 100

print("Accuracy :", accuracy, "%")
```
