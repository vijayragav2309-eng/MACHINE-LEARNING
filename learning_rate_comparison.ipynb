```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

# Small dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 1, 1, 0])   # XOR output


# -------------------------------
# Neural Network - Low Learning Rate
# -------------------------------
low_lr = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='relu',
    solver='adam',
    learning_rate_init=0.001,
    max_iter=5000,
    random_state=42
)

low_lr.fit(X, y)

low_pred = low_lr.predict(X)
low_accuracy = accuracy_score(y, low_pred)

print("LOW LEARNING RATE")
print("Predicted Output:", low_pred)
print("Accuracy:", low_accuracy)
print("Number of Iterations:", low_lr.n_iter_)


# -------------------------------
# Neural Network - High Learning Rate
# -------------------------------
high_lr = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='relu',
    solver='adam',
    learning_rate_init=0.1,
    max_iter=5000,
    random_state=42
)

high_lr.fit(X, y)

high_pred = high_lr.predict(X)
high_accuracy = accuracy_score(y, high_pred)

print("\nHIGH LEARNING RATE")
print("Predicted Output:", high_pred)
print("Accuracy:", high_accuracy)
print("Number of Iterations:", high_lr.n_iter_)


# -------------------------------
# Compare Loss
# -------------------------------
plt.plot(low_lr.loss_curve_, label="Low LR = 0.001")
plt.plot(high_lr.loss_curve_, label="High LR = 0.1")

plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.title("Low vs High Learning Rate")
plt.legend()
plt.show()
```
